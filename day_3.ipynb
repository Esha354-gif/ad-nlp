```python
# Creating a List with 5 elements
my_list = [10, 20, 30, 40, 50]

# Accessing elements from the list by index
print(f"First element of List: {my_list[0]}")   # Accessing the 1st element
print(f"Third element of List: {my_list[2]}")   # Accessing the 3rd element
print(f"Last element of List: {my_list[-1]}")   # Accessing the last element using negative indexing

# Creating a Tuple with 5 elements
my_tuple = (5, 15, 25, 35, 45)

# Accessing elements from the tuple by index
print(f"Second element of Tuple: {my_tuple[1]}") # Accessing the 2nd element
print(f"Fourth element of Tuple: {my_tuple[3]}") # Accessing the 4th element
print(f"Last element of Tuple: {my_tuple[-1]}")  # Accessing the last element using negative indexing

# Creating a Dictionary with 5 key-value pairs
my_dict = {'a': 100, 'b': 200, 'c': 300, 'd': 400, 'e': 500}

# Accessing values from the dictionary using keys
print(f"Value associated with key 'a': {my_dict['a']}")
print(f"Value associated with key 'c': {my_dict['c']}")
print(f"Value associated with key 'e': {my_dict['e']}")
```
